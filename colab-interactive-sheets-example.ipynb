```python
from google.colab import auth
from google.auth import default
import gspread
from google.colab import sheets
import pandas as pd
import numpy as np
from vega_datasets import data
df = data.us_state_capitals()
df
```

```python
import altair as alt
alt.renderers.enable('colab')
```

```python
sheet = sheets.InteractiveSheet(title="State Captitals", include_column_headers=True, df=df)
df2 = sheet.as_df()
df2
```

```python
import altair as alt

# Create a base map of the US states
states = alt.topo_feature(data.us_10m.url, 'states')

# Define the base chart for the map
base = alt.Chart(states).project('albers').properties(
    title='US State Capitals'
)

# Create the state borders layer
state_borders = base.mark_geoshape(
    fill='lightgray',
    stroke='white'
)

# Create the capitals layer
capitals = alt.Chart(df2).mark_circle().encode(
    longitude=alt.X('lon', type='quantitative'),
    latitude=alt.Y('lat', type='quantitative'),
    size=alt.value(50), # Adjust size of points
    color=alt.value('red'), # Color the points red
    tooltip=['state', 'city'] # Add tooltips for state and city
).properties(
    title='State Capitals'
)

# Combine the state borders and capitals layers and set overall chart size
map_chart = (state_borders + capitals).properties(
    width=800,  # Set a larger width
    height=500  # Set a larger height
).interactive() # Add interactivity for zoom and pan

# Display the map
map_chart
```
