```python
import pandas as pd
import os

# Note: This requires kaggle.json to be uploaded to /content/kaggle.json or configured in secrets
# For this demonstration, I will attempt to download it using the kaggle CLI.
!pip install -q kaggle
if not os.path.exists('netflix_titles.csv'):
    !kaggle datasets download -d shivamb/netflix-shows --unzip

df = pd.read_csv('netflix_titles.csv')
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Filter for movies only
movies_df = df[df['type'] == 'Movie']

# Group by release year and count
movies_per_year = movies_df['release_year'].value_counts().sort_index()

# Plotting
plt.figure(figsize=(12, 6))
sns.lineplot(x=movies_per_year.index, y=movies_per_year.values, marker='o', color='red')
plt.title('Number of Movies Released per Year on Netflix')
plt.xlabel('Release Year')
plt.ylabel('Number of Movies')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
